```python
df = pd.read_csv("../project/Doutorado Alessandro Macêdo/TTC_C_x_13M_D2_0001.csv")
df = df[(df["x_i"].between(3,36)) & (df["x_j"].between(3,36))]
df["time"] = df["time"]/(30*60)
df["risk"] = ((1.5-df["TTC"])**2) * df["vx_i"].abs() * df["vx_j"].abs() * df["vy_i"].abs() * df["vy_j"].abs()
```

```python
df[df["TTC"].between(0,1.5)].sort_values("risk",ascending=False)[["track_id_i","vehicle_type_i","track_id_j","vehicle_type_j","time","TTC","risk"]].drop_duplicates(subset=["track_id_i","track_id_j"]).head(50) # 
```
